```python
#IMPORTING THE INIT_CHAT_MODEL TO INITIALIZE OUR MODEL WE ARE USING IN THIS 
from langchain.chat_models import init_chat_model
```

```python
#IMPORTING HUMAN AI SYSTEM MESSAGES TO CREATE THE MEMORY FOR THE MODEL SO IT CAN RECOGNIZE WHICH MESSAGE IS SPOKEN BY WHOM AI OR HUMAN
from langchain_core.messages import (HumanMessage, AIMessage, SystemMessage)
```

```python
#INSTALLING THIS TO USE MY MISTRAL MODEL IN LANGCHAIN
!pip install langchain-mistralai
```

```python
#INITIALIZING MY MISTRAL MODEL AND API
llm = init_chat_model(
    "mistral-large-latest",
    model_provider="mistralai",
    api_key="h5gcvuLKeMh9VEX872i5HygOas6EC6HU"
)
```

```python
#CREATING THE HISTORY FOR MY MODEL
chat_history = []
```

```python
#DESIGN THE PROMPT AND SETTING THE BEHAVIOUR OF MY MODEL BY USING SYSTEM MESSAGE
system_message = SystemMessage(
    content="""
You are a helpful medical assistant chatbot.
Rules:
- Give simple and safe health information
- Do NOT give dangerous or exact prescriptions
- Always suggest consulting a doctor for serious issues
- Be friendly and clear
- Return the clean response dont add any special characters
- Return answer in bullet points and small summary
- Dont add these special characters ** Use proper bold fonts for headings
"""
)
```

```python
#ASKING THE USER TO WRITE HIS MED QUERY AND OUR CHATBOT WILL ANSWER HIM/HER WHILE SAVING THE CHAT HISTORY TOO TO EXIT THE CHAT WE WRITE EXIT
while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Chat ended.")
        print("Your chat : " , chat_history )
        break
        
    chat_history.append(HumanMessage(content=user_input))

    message = [system_message] + chat_history

    response = llm.invoke(message)

    chat_history.append(AIMessage(content=response.content))

    print(response.content)
```
