```python
#IMPORTING THE MATPLOTLIB AND SEABORN TO VISUALIZE OUR DATA
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
#USING SCATTER PLOT TO SEE THE RELATIONS BTW FEATURES FIRSTLY IM SEEING THE REL OF OUTER PART OF THE FLOWER
sns.scatterplot(x="SepalWidthCm",y="SepalLengthCm",hue="Species",data=df)
plt.show()
```

```python
#NOW IM SEEING THE REL OF INNER PART OF THE FLOWER
sns.scatterplot(x="PetalWidthCm",y="PetalLengthCm",hue="Species",data=df)
plt.show()
```

```python
#EXTRACTING ONLY NUMERIC COLUMNS AND CHECKING THERE CORR RELATIONSHIP
df.corr(numeric_only=True)
```

```python
#SEEING THE DISTRIBUTION OF EACH NUMERIC COL BY USING HISTPLOT FROM WHICH WE CAN SEE THE SKEWNESS OF OUR DATA
for c in df.columns:
    sns.histplot(x=c,data=df,kde=True)
    plt.show()
```

```python
#SPECIES WISE
for c in df.columns:
    sns.histplot(x=c,data=df,hue="Species",kde=True,element="step")
    plt.show()
```

```python
num = df.drop(["Species"],axis=1)
```

```python
num
```

```python
#CHECKING THE OUTLIERS OF THE NUM COL BY USING BOXPLOT
for c in num.columns:
    sns.boxplot(num[c])
    plt.show()
```

```python
#CHEKING THE SKEWNESS BY NUMERIC VALUES
for c in num.columns:
    print(f"{c} mean : {num[c].mean()} median : {num[c].median()} skew : {num[c].skew()}")
```
