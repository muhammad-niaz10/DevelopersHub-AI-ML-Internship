```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
numerical_pipeline = Pipeline([
    ("impute",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler())
])
```

```python
cat_pipeline = Pipeline([
    ("impute",SimpleImputer(strategy="most_frequent")),
    ("encoder",OneHotEncoder())
])
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer([
    ("numeric",numerical_pipeline,num_cols),
    ("cat",cat_pipeline,cat_cols)
])
```

```python
from sklearn.tree import DecisionTreeClassifier

pipe = Pipeline([
    ("preprocessor",preprocessor),
    ("model",DecisionTreeClassifier())
])
```

```python
pipeline = joblib.load("pipe.pkl")
```

```python
def get_user_input():
    data = {
        "age": float(input("Age: ")),
        "sex": input("Sex (Male/Female): "),
        "cp": input("Chest Pain Type: "),
        "trestbps": float(input("Resting BP: ")),
        "chol": float(input("Cholesterol: ")),
        "fbs": input("FBS (True/False): "),
        "restecg": input("Rest ECG: "),
        "thalch": float(input("Max Heart Rate: ")),
        "exang": input("Exercise Angina (True/False): "),
        "oldpeak": float(input("Oldpeak: ")),
        "slope": input("Slope: "),
        "ca": input("CA: "),
        "thal": input("Thal: ")
    }

    df = pd.DataFrame([data])

    df["fbs"] = df["fbs"].map({"True": True, "False": False})
    df["exang"] = df["exang"].map({"True": True, "False": False})

    return df
```

```python
import shap

model = pipeline.named_steps["model"]

preprocessor = pipeline.named_steps["preprocessor"]

X_train_transformed =  preprocessor.transform(X_train)

explainer = shap.TreeExplainer(model,X_train_transformed)

featured_columns = preprocessor.get_feature_names_out()
```

```python
def predict_disease(pipeline):
    user_df = get_user_input()

    prediction = pipeline.predict(user_df)[0]

    print("\nPrediction:", prediction)

    if prediction == 1:
        print("⚠ High risk of heart disease")
    else:
        print("✅ Low risk")

    user_transform = preprocessor.transform(user_df)
    shap_values = explainer.shap_values(user_transform)

    print("\n🔍 SHAP Explanation:")

    print(shap_values.shape)

    shap_array = shap_values[0]

    values = shap_array[:, 1]

    feature_names = pipeline.named_steps["preprocessor"].get_feature_names_out()

    exp = shap.Explanation(
    values=values,
    data=user_transform[0],
    feature_names=feature_names
)

    shap.plots.bar(exp)

    sorted_data = sorted(
        zip(feature_names, values),
        key=lambda x: abs(x[1]),
        reverse=True
    )

    for name, val in sorted_data:
        print(f"{name}: {val}")
```

```python
predict_disease(pipeline)
```
